```python
import requests
import os
import time
from urllib.parse import quote
import re

def download_images_bing(query, save_folder, num=100):
    os.makedirs(save_folder, exist_ok=True)
    
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    }
    
    downloaded = 0
    offset = 0
    
    while downloaded < num:
        url = f"https://www.bing.com/images/search?q={quote(query)}&count=35&offset={offset}&mkt=en-US"
        
        try:
            response = requests.get(url, headers=headers, timeout=15)
            
            # Extract image URLs
            pattern = r'"murl":"(https?://[^"]+\.(?:jpg|jpeg|png|JPG|JPEG|PNG))"'
            urls = re.findall(pattern, response.text)
            
            if not urls:
                # Try alternative pattern
                pattern2 = r'imgurl=([^&]+\.(?:jpg|jpeg|png))&'
                urls = re.findall(pattern2, response.text)
            
            if not urls:
                print(f"  No more images found at offset {offset}")
                break
                
            for img_url in urls:
                if downloaded >= num:
                    break
                try:
                    img_url = requests.utils.unquote(img_url)
                    img_response = requests.get(
                        img_url, 
                        headers=headers, 
                        timeout=10,
                        stream=True
                    )
                    
                    if img_response.status_code == 200:
                        content_type = img_response.headers.get('content-type', '')
                        if 'image' in content_type:
                            filename = f"{save_folder}/{downloaded+1:04d}.jpg"
                            with open(filename, 'wb') as f:
                                f.write(img_response.content)
                            downloaded += 1
                            
                            if downloaded % 20 == 0:
                                print(f"  {downloaded} images downloaded...")
                except:
                    continue
                    
            offset += 35
            time.sleep(1)
            
        except Exception as e:
            print(f"  Error: {e}")
            break
    
    return downloaded

# =============================================
# DOWNLOAD ALL 4 CLASSES
# =============================================

classes = {
    "mouse":    "computer mouse",
    "keyboard": "computer keyboard",
    "phone":    "smartphone android",
    "charger":  "mobile phone charger"
}

for class_name, query in classes.items():
    print(f"\n{'='*40}")
    print(f"Downloading: {class_name.upper()}")
    print(f"{'='*40}")
    
    folder = f"dataset/{class_name}"
    count  = download_images_bing(query, folder, num=100)
    
    print(f"DONE! {count} images saved in dataset/{class_name}/")

print("\n" + "="*40)
print("ALL 4 CLASSES DOWNLOADED!")
print("="*40)
```

```python
from ddgs import DDGS
import requests
import os
import time

def download_more(query, save_folder, target=150):
    os.makedirs(save_folder, exist_ok=True)
    current = len(os.listdir(save_folder))
    
    if current >= target:
        print(f"  Already have {current} images — skipping!")
        return current
    
    needed = target - current
    print(f"  Have {current}, need {needed} more...")
    
    downloaded = current
    results    = DDGS().images(query, max_results=needed + 30)
    
    for result in results:
        if downloaded >= target:
            break
        try:
            response = requests.get(result["image"], timeout=10)
            if response.status_code == 200:
                filepath = f"{save_folder}/{downloaded+1:04d}.jpg"
                with open(filepath, "wb") as f:
                    f.write(response.content)
                downloaded += 1
        except:
            continue
    
    time.sleep(2)
    return downloaded

# Different search terms to get fresh unique images
extra_searches = {
    "mouse": [
        "wireless computer mouse white background",
        "gaming mouse rgb",
        "logitech mouse desk",
        "computer mouse top view"
    ],
    "keyboard": [
        "mechanical keyboard rgb",
        "wireless keyboard white",
        "office keyboard desk",
        "gaming keyboard setup"
    ],
    "phone": [
        "android smartphone front view",
        "mobile phone screen on",
        "samsung galaxy phone",
        "iphone on table"
    ],
    "charger": [
        "USB type c charger",
        "white phone charger plug",
        "mobile charging adapter",
        "phone charger wall plug"
    ]
}

for class_name, queries in extra_searches.items():
    print(f"\nTopping up: {class_name.upper()}")
    print(f"{'='*40}")
    
    for query in queries:
        current = len(os.listdir(f"dataset/{class_name}"))
        if current >= 150:
            print(f"  Reached 150 — moving to next class!")
            break
        print(f"  Searching: '{query}'")
        count = download_more(query, f"dataset/{class_name}", target=150)
        time.sleep(3)

# Final count
print("\n" + "="*40)
print("FINAL COUNT:")
print("="*40)
for class_name in ["mouse", "keyboard", "phone", "charger"]:
    count  = len(os.listdir(f"dataset/{class_name}"))
    status = "✅ Good" if count >= 100 else "❌ Need more"
    print(f"  {class_name:12} : {count:3} images — {status}")
```
